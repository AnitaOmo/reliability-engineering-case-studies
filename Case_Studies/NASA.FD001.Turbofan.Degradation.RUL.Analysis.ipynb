```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Load the data exactly like we did before
columns = ['engine_id', 'cycle', 'setting_1', 'setting_2', 'setting_3'] + [f'sensor_{i}' for i in range(1, 22)]
data = pd.read_csv(r"C:\Users\HP\OneDrive\Documents\test_FD001.txt", sep=r'\s+', header=None, names=columns)

# 2. Isolate Engine #1
engine_1 = data[data['engine_id'] == 1]

# 3. Plot Sensor 2 over its 31 cycles
plt.figure(figsize=(9, 4))
plt.plot(engine_1['cycle'], engine_1['sensor_2'], marker='o', color='blue', linewidth=2)

plt.xlabel('Time (Cycles)', fontsize=11)
plt.ylabel('HPC Outlet Temperature (Sensor 2)', fontsize=11)
plt.title('Engine #1: Live Sensor Degradation Tracking', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()
```
